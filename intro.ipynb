```python
def search_text(text_search: str, list_text: list[str]) -> int | None:
    index_found = None
    for i, text in enumerate(list_text):
        if text_search == text:
            index_found = i
            break
    return index_found
```

```python
assert search_text('Brest', cities) == 3
assert search_text('Pau', cities) is None  # is not None
```

```python
# exo : chercher les villes Brest, brest et Pau dans la liste cities et afficher pour chacune si elle présente ou pas
# avec un joli affichage
all_cities = ['Marseille', 'Nantes', 'Biarritz', 'Brest', 'Rennes']
cities_searched = ['Brest', 'brest', 'Pau']

for city_search in cities_searched:
    index_found = search_text(city_search, all_cities)
    if index_found is not None:
        print(f"{city_search} found at index {index_found}")
    # elif True == False:
    #     pass
    else:
        print(f"{city_search} not found")
```

```python
def search_text(text_search: str, list_text: list[str], case: bool = True) -> int | None:
    index_found = None
    text_search_c = text_search if case else text_search.casefold()
    for i, text in enumerate(list_text):
        text_c = text if case else text.casefold()
        if text_search_c == text_c:
            index_found = i
            break
    return index_found
```

```python
assert search_text('Brest', all_cities) == 3
assert search_text('brest', all_cities) is None
assert search_text('brest', all_cities, case=False) == 3
```
