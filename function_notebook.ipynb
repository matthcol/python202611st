## Mes fonctions

```python
def f(a, b, *args, c=0, d=None, e=[], **kwargs):
    print("a:", a)
    print("b:", b)
    print("args:", args)
    print("c:", c)
    print("d:", d)
    print("e:", e)
    print("kwargs:", kwargs)
    print()
```

```python
# f()   TypeError: f() missing 2 required positional arguments: 'a' and 'b'
f(1, "toto")
f(1, "toto", "tram", "bus", "covoiturage", mois="novembre", c=5, d=7, e=[12, 13], saison="été")
f(1, "toto", "tram", "bus", "covoiturage", mois="novembre", c=5, d=7, saison="été")
# f(1, "toto", "tram", "bus", "covoiturage", mois="novembre", c=5, d=7, saison="été", mois="décembre") # SyntaxError: keyword argument repeated: mois
```

```python
values = [13, 64, 44]
options = {"transport": "inconnu", "destination": "Tarbes", "d": 123}
f(*values, **options)
```

## str vs repr
- str : public = end user
- repr : public = dev

```python
city = 'Marseille'
print(city) # calls str()
print(str(city))
print(repr(city))
city # calls repr()
```

```python
class Person:
    pass

p = Person()
print(p)
p
```

```python
class Person:
    def __init__(self, name: str) -> None:
        self.name = name

    # both repr() / str()
    def __repr__(self) -> str:
        return f"Person(name={self.name})"

p = Person("John Doe")
print(p)
p
```

```python
class Person:
    def __init__(self, name: str) -> None:
        self.name = name

    # repr() par défaut
    
    # str()
    def __str__(self) -> str:
        return f"Person(name={self.name})"

p = Person("John Doe")
print(p)
p
```

```python
class Person:
    def __init__(self, name: str) -> None:
        self.name = name

    # repr()
    def __repr__(self) -> str:
        return f"Person(name={self.name})"
    
    # str()
    def __str__(self) -> str:
        return f"{self.name}"

p = Person("John Doe")
print(p)
p
```

```python
class Resource:

    def __enter__(self):
        print('Resource enter')
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print('Resource exit')
```

```python
with Resource() as r:
    print("Use resource")
```

```python
with Resource() as r:
    print("Use resource")
    raise ValueError("Error in the with")
```
